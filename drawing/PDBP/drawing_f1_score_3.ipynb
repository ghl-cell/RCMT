```python
with open('CTMT.txt', 'r', encoding='utf-8', errors='replace') as f:
    log_content = f.read()
```

```python
Val_F1_pattern = r'val f1:(\d+\.\d+)'
Val_F1_RNA_pattern = r'val f1_T:(\d+\.\d+)'
Val_F1_MRI_pattern = r'val f1_M:(\d+\.\d+)'
rate_pattern = r'rate:(\d+\.\d+)'


Val_F1 = re.findall(Val_F1_pattern, log_content)
Val_F1_RNA = re.findall(Val_F1_RNA_pattern, log_content)
Val_F1_MRI = re.findall(Val_F1_MRI_pattern, log_content)
rate_CTMT = re.findall(rate_pattern, log_content)

# 转换为数值类型
Val_F1 = [float(f_1) for f_1 in Val_F1]
Val_F1_RNA = [float(f_1) for f_1 in Val_F1_RNA]
Val_F1_MRI = [float(f_1) for f_1 in Val_F1_MRI]
rate_CTMT = [float(f_1) for f_1 in rate_CTMT]

# 生成epoch编号
epochs_CTMT = list(range(1, len(Val_F1) + 1))

print(f"共找到 {len(Val_F1_RNA)} 个epoch的训练数据")
print("train_F1_RNA:", Val_F1_RNA[:10], "...")  # 显示前10个
print("train_F1_MRI:", Val_F1_MRI[:10], "...")     # 显示前10个
```

```python
with open('RNA.txt', 'r', encoding='utf-8', errors='replace') as f:
    log_content = f.read()
```

```python
Val_F1_RNA_single_pattern = r'val f1:(\d+\.\d+)'


Val_F1_RNA_single = re.findall(Val_F1_RNA_single_pattern, log_content)


# 转换为数值类型
Val_F1_RNA_single = [float(f_1) for f_1 in Val_F1_RNA_single]


# 生成epoch编号
epochs_RNA = list(range(1, len(Val_F1_RNA_single) + 1))

print(f"共找到 {len(Val_F1_RNA_single)} 个epoch的训练数据")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import savgol_filter

def savgol_smooth(data, window_length=11, polyorder=3):
    """Savitzky-Golay滤波器，处理边界更好"""
    return savgol_filter(data, window_length, polyorder, mode='nearest')
```

```python
# 对数据进行平滑处理
window_length = 17  # 窗口长度（奇数）
polyorder = 3       # 多项式阶数
# 对准确率数据进行平滑
# Val_F1_RNA_single_smooth = savgol_smooth(Val_F1_RNA_single, window_length, 2)
Val_F1_RNA_smooth = savgol_smooth(Val_F1_RNA, window_length, 2)
Val_F1_RNA_concat_smooth = savgol_smooth(Val_F1_RNA_concate, window_length, 2)


# Val_F1_MRI_single_smooth = savgol_smooth(Val_F1_MRI_single, window_length, 2)
Val_F1_MRI_smooth = savgol_smooth(Val_F1_MRI, window_length, 2)
Val_F1_MRI_concat_smooth = savgol_smooth(Val_F1_MRI_concate, window_length, 2)



Val_F1_smooth = savgol_smooth(Val_F1, window_length, 2)
Val_F1_concat_smooth = savgol_smooth(Val_F1_concate, window_length, 2)

rate_CTMT_smooth = savgol_smooth(rate_CTMT, window_length, 2)
rate_concate_smooth = savgol_smooth(rate_concate, window_length, 2)

Val_F1_RNA_single_smooth = savgol_smooth(Val_F1_RNA_single, window_length, 2)
Val_F1_MRI_single_smooth = savgol_smooth(Val_F1_MRI_single, window_length, 2)
```

```python
import matplotlib.pyplot as plt

# 绘制曲线
plt.figure(figsize=(6, 5))

plt.plot(epochs_concat[:100], Val_F1_RNA_single_smooth[::70], color='#edb120',linestyle='--', markersize=5, linewidth=2, label='Single RNA')
plt.plot(epochs_concat[:100], Val_F1_RNA_concat_smooth[::2]*0.9, color='#0072bd', markersize=5, linewidth=2, label='RNA in MM')
plt.plot(epochs_concat[:100], Val_F1_RNA_smooth[::6], color='#d95319', markersize=5, linewidth=2, label='RNA in CTMT')

# plt.plot(epochs, train_losses_oGL_smooth, color='#edb120', marker='D', markersize=5, linewidth=2, label='w/o GL')
# plt.plot(epochs, train_losses_oDementia_smooth, color='#edb120', marker='D', markersize=5, linewidth=2, label='w/o Dementia')
plt.xlabel("Training Epoch",fontsize=18)
plt.ylabel("Val F1-score",fontsize=18)
plt.xlim(1, len(epochs_concat[:100]))         # 横轴范围从1到最大epoch
# plt.ylim(0.6, 1.0)               # y轴范围
plt.xticks(range(1, len(epochs_concat[:100])+1, max(len(epochs_concat[:100])//10, 1)),fontsize=16)
plt.yticks(fontsize=16)
# plt.title("Training Loss Curve")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=16)
plt.tight_layout()
plt.savefig('PDBP-RNA.pdf',dpi=600, bbox_inches='tight')
plt.savefig('PDBP-RNA.png',dpi=600, bbox_inches='tight')
```
