```python
PPMI_all_train_f1_CT=pd.read_csv('PPMI/f1-score-data/all_train_f1_CT.csv').drop(['Unnamed: 0'],axis=1)
PPMI_all_train_f1_T_CT=pd.read_csv('PPMI/f1-score-data/all_train_f1_T_CT.csv').drop(['Unnamed: 0'],axis=1)
PPMI_all_train_f1_M_CT=pd.read_csv('PPMI/f1-score-data/all_train_f1_M_CT.csv').drop(['Unnamed: 0'],axis=1)
PPMI_ratio_list_CT=pd.read_csv('PPMI/f1-score-data/ratio_list_CT.csv').drop(['Unnamed: 0'],axis=1)

PPMI_all_train_f1_CTO=pd.read_csv('PPMI/f1-score-data/all_train_f1_CTO.csv').drop(['Unnamed: 0'],axis=1)
PPMI_all_train_f1_T_CTO=pd.read_csv('PPMI/f1-score-data/all_train_f1_T_CTO.csv').drop(['Unnamed: 0'],axis=1)
PPMI_all_train_f1_M_CTO=pd.read_csv('PPMI/f1-score-data/all_train_f1_M_CTO.csv').drop(['Unnamed: 0'],axis=1)
PPMI_ratio_list_CTO=pd.read_csv('PPMI/f1-score-data/ratio_list_CTO.csv').drop(['Unnamed: 0'],axis=1)
```
