```python
import numpy as np
import matplotlib.pyplot as plt

# 数据准备
metrics = ['F1-score','Accuracy','Precision','Recall','AUROC','AUPR']
# methods = ['SVM', 'RF', 'XGBoost', 'MADDI', 'MOGONET', 'Ours']
methods = ['MM-Transformer', 'MADDI', 'MOGONET', 'Ours']
data = {
    #transformer=[0.786,0.812,0.748,0.763,0.904,0.821,0.786]
    # maddi = [0.810,0.849,0.936,0.745,0.885,0.836,0.810]
    # mogonet=[0.728,0.794,0.734,0.729,0.908,0.783,0.728]
    # ours=[0.886,0.934,0.958,0.841,0.983,0.938,0.886]
    'MM-Transformer': np.concatenate((PPMI_result[PPMI_result['Methods']=='Transformer'].values[0][1:],PPMI_result[PPMI_result['Methods']=='Transformer'].values[0][1:3])).tolist()[:-1],
    'MADDI': np.concatenate((PPMI_result[PPMI_result['Methods']=='MADDI'].values[0][1:],PPMI_result[PPMI_result['Methods']=='MADDI'].values[0][1:3])).tolist()[:-1],
    'MOGONET': np.concatenate((PPMI_result[PPMI_result['Methods']=='MOGONET'].values[0][1:],PPMI_result[PPMI_result['Methods']=='MOGONET'].values[0][1:3])).tolist()[:-1],
    'Ours': np.concatenate((PPMI_result[PPMI_result['Methods']=='Ours'].values[0][1:],PPMI_result[PPMI_result['Methods']=='Ours'].values[0][1:3])).tolist()[:-1]
}

# 绘制雷达图
fig, axs = plt.subplots(2, 2, figsize=(10, 10), subplot_kw=dict(polar=True))
axs = axs.flatten()
colors=['#004E66','#51A3BC','#FCBE32','#FF5F2E']
for i, ax in enumerate(axs):
    method1 = methods[i]
    values1 = data[method1]
    
    angles = np.linspace(0, 2 * np.pi, len(metrics), endpoint=False).tolist()
    angles += angles[:1]
    # angles += angles[:1]

    ax.fill(angles, values1, color=colors[i], alpha=0.25)
    ax.plot(angles, values1, color=colors[i], linewidth=2, label=method1)
    ax.plot(angles, data['Ours'], color=colors[-1], linewidth=2)


    ax.set_rgrids(np.linspace(0.2, 1.0, num=5), labels=np.linspace(0.2, 1.0, num=5), angle=0)
    ax.set_yticklabels([], fontsize=10)
    ax.set_thetagrids(np.degrees(angles[:-1]), metrics, fontsize=18)
    ax.set_title(f'{method1}',fontdict={'size': 20})
    # ax.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))

plt.tight_layout()
# plt.show()
plt.savefig('Fig8_PPMI_radar.png',dpi=600)
```
